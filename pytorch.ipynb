```python
import torch

# 随机张量
Y = torch.rand((5,3))  #
print(Y)

# 服从某种分布的张量
z = torch.normal(mean=0.0,std=torch.tensor(1.0)) # 均值0，方差1
c = torch.normal(mean=0.0,std=torch.rand(2)) # 均值0，方差1
print(z)
print(c)

X = torch.randn(5, 3)
print(X)

# Tensor类的构造函数，数据类型默认为float32
a = torch.tensor([1,2,3,4,5])
x = torch.Tensor(a)
print(x)


# 特殊张量
print(torch.ones_like(X,dtype=torch.int64))   #单位矩阵
print(torch.eye(4,1))

print(x.device)

#稀疏张量  layout
indices = torch.tensor([[0,1,1],[2,0,2]])  #定义了三对坐标
values = torch.tensor([1,2,3]) # 值分别是1，2，3
x = torch.sparse_coo_tensor(indices, values,(2,4)).to_dense()  #转为稠密张量
print(x)

```
